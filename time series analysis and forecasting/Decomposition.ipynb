```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# Loading airpassengers dataset

# url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv'
# url ='AirPassengers.csv'

# df = pd.read_csv('AirPassengers.csv', header=0, index_col = 0, parse_dates=True, squeeze=True)
# df = pd.read_csv(url, header=0, index_col = 0, parse_dates=True, squeeze=True)


# OR

# Loading airpassengers dataset
# url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv'
df = pd.read_csv('AirPassengers.csv', header=0, index_col=0, parse_dates=True)

# If you still want a Series instead of a DataFrame (which is what 'squeeze=True' did)
# you can convert it after reading:
if len(df.columns) == 1:
    df = df.iloc[:, 0]  # This converts the DataFrame to a Series if it has only one column
```

Mulplicative part

```python
# Perform seasonal decompose
result = seasonal_decompose(df, model='multiplicative')
```

```python
# Plot the components
plt.rcParams.update({'figure.figsize': (10,10)})
result.plot()
plt.show()
```

Additive part

```python
# Perform seasonal decompose
result2 = seasonal_decompose(df, model='additive')
```

```python
# Plot the components
plt.rcParams.update({'figure.figsize': (10,10)})
result2.plot()
plt.show()
```
