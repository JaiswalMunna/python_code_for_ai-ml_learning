```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('50_startups.csv')
df.info()
```

```python
df.head(10)
```

```python
#Dividing the data into dependents and indepenedents which is Y-Var and X-Var

X = df.iloc[:,0].values  #[:,0] --->  R&D Spend
Y = df.iloc[:,-1].values #[:,-1] --> Profit
```

```python
X
```

```python
# train test split 

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2) #test_size=0.2 means we have 50 data. out of 50 data 0.2(20%) data willbe used for test and reamining 80% data used for train 
```

```python
X_train = X_train.reshape(-1,1)
X_test = X_test.reshape(-1,1)
```

```python
X_train
```

```python
X_test
```

```python
#Feature Scaling 

from sklearn.preprocessing import StandardScaler

sc=StandardScaler()

X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
#creating linear regression model

from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, Y_train)
```

```python
#predictions

Y_pred= regressor.predict(X_test)
```

```python
#ploting of y_test vs y_pred

plt.plot(Y_test,color='blue',label='test')
plt.plot(Y_pred,color='red',label='predictions')
plt.show()
```

Regression Metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(Y_pred,Y_test)
print(mae)
```

```python
mse = mean_squared_error(Y_pred,Y_test)
print(mse)
```

```python
print(np.sqrt(mse))
```

```python
r2 = r2_score(Y_pred,Y_test)
print(r2)
```

```python
#Calculated the adjusted r squared

n=X_test.shape[0]
k=X_test.shape[1]

adjusted_r2=1-(1-r2)*(n-1)/(n-1-k)
print(adjusted_r2)
```
